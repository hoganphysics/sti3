```python
from stipy import *
```

```python
testDevice = connect("localhost/0/TestDevice", "192.168.1.109:2809")
```

```python
lm=testDevice.getLogManager()
```

```python
dir(lm)
```

```python
from datetime import datetime
# import pandas as pd  # optional, just for date formatting if you want

today = datetime.now().strftime("%Y/%m/%d")

f = LogFileFilter()
f.logName = "testing"     # replace with one of lm.getLogNames()
f.startDate = today
f.endDate = today
f.startIndex = 0
f.endIndex = -1

ids = lm.getLogIDs(f)
ids
```

```python
log_id = ids[-1]          # usually the newest index for that day
browser = lm.openLog(log_id, tail_lines=10)
print(browser)
```

```python
browser.page_backward(50)
print(browser.text)
```

```python
browser.page_forward(200)
print(browser.text)
```

```python
browser.save_local("downloaded_test_log.log")
```
