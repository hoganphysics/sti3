# Image read

Start the readWrite example device, read one image channel, and open it with Pillow.

```python
from pathlib import Path
import subprocess
import sys
import time

import stipy
from PIL import Image as PILImage


name_service = "192.168.88.252:2809"
device_process = subprocess.Popen([sys.executable, "testDevice.py"])
time.sleep(2.0)

device = stipy.connect(stipy.DeviceID("TestDevice", "localhost", 0), name_service)
device
```

Channel 15 returns a file-backed `Image`. Transfer it into Python memory and decode it with Pillow.

```python
image = device.read(15)
pil_image = image.to_pil(device)
pil_image
```

Transfer the same image to a local file, then open that file with Pillow.

```python
image_path = Path("image_read.png")
image.to_file(image_path, device)

pil_image_from_file = PILImage.open(image_path)
pil_image_from_file.load()
pil_image_from_file
```

```python
device_process.terminate()
```
