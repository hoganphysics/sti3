```python
did = DeviceID("STI Server", "localhost", 0)
server = connect(did, "192.168.1.6:2809")
```

```python
server.getID()
```

```python
def f():

    setvar("debug", True)

    d1 = dev("Device1", "localhost", 0)
    d2 = dev("Device2", "localhost", 0)
    d3 = dev("Device3", "localhost", 0)
    # c0 = ch(d1, 0)

    g = group("MOT")

    ms = 1000000
    # event(ch(checker, 0), 1, 1.01)
    time=100*ms
    event(ch(d1, 0), time, 1.1)
    event(ch(d1, 1), time, 1.2)
    event(ch(d3, 0), time, 4.5)
    time+=1000*ms
    event(ch(d1, 0), time, 2.31)
    time += 1
    event(ch(d1, 1), time, 3.4)
    time += 1
    g.event(ch(d1, 0), time, 2.5)
    g.event(ch(d3, 0), time, 7.5)
    g.setvar("test", 1234)
```

```python
s=server.makeshot(f)
```

```python
gflat=s.rootgroup().flatten()
gflat[1].vars()
```

```python
g2=RawEventGroup.unflatten(gflat)
```
