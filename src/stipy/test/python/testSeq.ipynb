```python
shots=[]
for key in s.sequenceTable.keys():
    print(s.sequenceTable[key].overwritten)
    shot = server.makeshot(s.shotmaker, s.sequenceTable[key].overwritten)
    shots.append(shot)

    seqEntryID = SequenceEntryID()
    seqEntryID.seqID = seq
    seqEntryID.seqIndex.index = key
    seqEntryID.seqIndex.repeat = 0

    parseTick = server.parse(shot, seqEntryID)
    parseTick.wait()

    resultTick = server.play(parseTick)
    resultTick.wait()
```
